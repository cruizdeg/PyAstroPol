#### MgF2 coated Quartz retarder

Current example is uncoated; coating to be added

```python
# %matplotlib inline
# %matplotlib tk
# %matplotlib qt5
import numpy as np
from matplotlib import pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
#
import PyAstroPol as pap
pap.roundOffDisplay(5)
```

```python
Wave = 0.6328
S = pap.Source(4, Clear=10)                                  # Source for ray analysis
Sd = pap.Source(4, Clear=10)                                  # Source for display
S.Wavelength = Sd.Wavelength = Wave
D = pap.Detector(100)                                           # Rotate the detector
D.translateOrigin(z=1000)                                       # Move the detector
#
```

```python
Dia, Thick = 100, 0.1
Surf1 = pap.Surface(Dia, n1='Air', n2='Quartz', Interference=True)
Surf2 = pap.Surface(Dia, n1='Quartz', n2='Air', Interference=True)
Surf2.translateOrigin(z=Thick)
Surf1.propagateRays(S)
Surf2.propagateRays(Surf1.tRays)
```

```python
Cover = pap.Material('Air')
Substrate = pap.Material('Air')
FM_Cover, ACover = pap.getBerremanFieldMatrixIsotropic(Cover.getRefractiveIndexAt(Wave), Surf1.Beta)
IFM_Cover = pap.getInverseMatrix(FM_Cover)
FM_Substrate, ASubstrate = pap.getBerremanFieldMatrixIsotropic(Substrate.getRefractiveIndexAt(Wave), Surf1.Beta)
#
IFM_Cover = pap.flattenMatrix(IFM_Cover)
FM_Substrate = pap.flattenMatrix(FM_Substrate)
```

```python
FM1 = Surf1.FieldMatrix
PM1 = pap.getBerremanPhaseMatrix(Surf1.Alpha, Thick)
IFM1 = pap.getInverseMatrix(FM1)
#
FM1 = pap.flattenMatrix(FM1)
PM1 = pap.flattenMatrix(PM1)
IFM1 = pap.flattenMatrix(IFM1)
#
CM1 = IFM_Cover*FM1*PM1*IFM1*FM_Substrate
```
